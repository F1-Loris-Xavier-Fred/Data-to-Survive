# Nettoyage de la Data Constructeur
## Initialisation

```python
import pandas as pd
import matplotlib as mpl

# On garde seulement les colonnes qui nous sont utiles pour alléger et optimiser
colonnes_voulues = ["constructorId","yearFrom","yearTo"]
df_constructors = pd.read_csv("../data/02_silver/f1db-constructors-chronology.csv", na_values=None, usecols=colonnes_voulues)

colonnes_voulues_2 = ["constructorId","constructorRef","name","nationality"]
df_constructors_2 = pd.read_csv("../data/02_silver/constructors.csv", na_values=None, usecols=colonnes_voulues_2)


display(df_constructors.shape) #test
display(df_constructors_2.shape) #test
display(df_constructors.dtypes) #test
display(df_constructors_2.dtypes) #test
```

## Filtre sur les années entre 1997 - 2026

Filtre sur les dates de fin `yearTo` null ou comprise entre 1997 et 2026 et date de début `yearFrom` inférieures ou égales à 2026.

```python
df_constructors = df_constructors[(df_constructors["yearTo"].isna()) | (df_constructors["yearTo"] >= 1997.0)]
df_constructors = df_constructors[df_constructors["yearFrom"] <= 2026]

df_constructors
```

## Récupération de la colonne id constructeur filtrée

```python
df_constructors["constructorId"] = df_constructors["constructorId"].str.replace("_", "", regex=False)
df_constructors["constructorId"] = df_constructors["constructorId"].str.replace("-", "", regex=False)
df_constructors["constructorId"] = df_constructors["constructorId"].str.replace(" ", "", regex=False)

#df_constructors = df_constructors["constructorId"]

df_constructors = df_constructors.rename(columns={
    "constructorId": "constructorRef"
})

df_constructors
```

## Normalisation de la colonne `constructorRef` du fichier 2 pour fusion

```python
df_constructors_2["constructorRef"] = df_constructors_2["constructorRef"].str.replace("_", "", regex=False)
df_constructors_2["constructorRef"] = df_constructors_2["constructorRef"].str.replace("-", "", regex=False)
df_constructors_2["constructorRef"] = df_constructors_2["constructorRef"].str.replace(" ", "", regex=False)

df_constructors_2
```

## Ajout des infos de `df_constructors_2` à la partie filtrée `df_constructors`
Maintenant qu'on a nettoyé les 2 tables et filtré la première sur la période qui nous intéresse, on ajouter les colonnes voulues de la 2 table

```python
df_ecurie = pd.merge(df_constructors, df_constructors_2, how="left", on=["constructorRef"])

df_ecurie = df_ecurie[["constructorId", "constructorRef", "name", "nationality"]]

df_ecurie = df_ecurie.sort_values("constructorRef", ascending=True)

df_ecurie = df_ecurie.rename(columns={
    "constructorId": "id_ecurie",
    "constructorRef": "ref_ecurie",
    "name": "nom_ecurie"
})

display(df_ecurie)
```
