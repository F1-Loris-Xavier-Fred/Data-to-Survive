# Nettoyage de la Data
## Initialistaion

```python
import pandas as pd
import matplotlib as mpl

# On garde seulement les colonnes qui nous sont utiles pour alléger et optimiser
colonnes_voulues = ["constructorId","yearFrom","yearTo"]
df_constructors = pd.read_csv("../data/02_silver/f1db-constructors-chronology.csv", na_values=None, usecols=colonnes_voulues)

display(df_constructors.shape) #test
display(df_constructors.dtypes) #test
```

## Filtre sur les années entre 1997 - 2026

Filtre sur les dates de fin `yearTo` null ou comprise entre 1997 et 2026 et date de début `yearFrom` inférieures ou égales à 2026.

```python
df_constructors = df_constructors[(df_constructors["yearTo"].isna()) | (df_constructors["yearTo"] >= 1997.0)]
df_constructors = df_constructors[df_constructors["yearFrom"] <= 2026]

df_constructors
```

## Renommage des colonnes
On renomme les colonnes pour qu'on se conforme au dictionnaire.

```python
df_circuits = df_circuits.rename(columns={
    "circuitId": "id_circuit",
    "circuitRef": "nom_circuit",
    "name": "nom_officiel",
})
```

## Filtrage par circuit
On garde seulement les 5 circuits qu'on a sélectionné pour notre étude, à savoir Monza, Silverstone, Monaco, Suzuka et Spa-Francorchamps.

```python
df_circuits = df_circuits[df_circuits["nom_circuit"].isin(["monza","spa","silverstone","monaco","suzuka"])]

print(df_circuits.shape)
df_circuits.head()
```

## Fusion de cette table avec traces_drapeaux_circuits
Maintenant qu'on a nettoyé cette table, on peut la fusionner avec la table qu'on a créé manuellement (donc propre)

```python
df_traces = pd.read_csv("../data/02_silver/traces_drapeaux_circuits.csv")

display(df_circuits.head())
display(df_traces.head())

df_fusion = (
    pd.merge(df_circuits, df_traces, how="cross")
    .loc[lambda d: d.apply(lambda r: r["nom_circuit"] in r["circuit"].lower(), axis=1)]
)
df_fusion = df_fusion[["id_circuit", "nom_circuit", "nom_officiel", "pays_code", "url_img_circuit"]]

df_fusion = df_fusion.rename(columns={
    "pays_code": "id_pays",
})

display(df_fusion.head())
```

## Sauvegarde du nettoyage
On sauvegarde le csv dans le dossier 03_gold

```python
df_fusion.to_csv("../data/03_gold/circuits.csv", index=False, encoding="utf-8")
```
