# Nettoyage de la Data
## Initialistaion

```python
import pandas as pd
import matplotlib as mpl

# On garde seulement les colonnes qui nous sont utiles pour alléger et optimiser
colonnes_voulues = ["statusId", "status"]
df_status = pd.read_csv("../data/02_silver/status.csv", usecols=colonnes_voulues)

df_status.shape #test
```

## Renommage des colonnes
On renomme les colonnes pour qu'on se conforme au dictionnaire.

```python
df_status = df_status.rename(columns={
    "statusId": "id_statu",
    "status": "libelle_statu",
})
```

## Nettoyage
On nettoie les libellés (espaces, valeurs nulles, doublons)

```python
# Nettoyage des libellés (espaces, valeurs nulles, doublons)
df_status["libelle_statu"] = df_status["libelle_statu"].astype(str).str.strip()
df_status = df_status.dropna(subset=["id_statu", "libelle_statu"])
df_status = df_status[df_status["libelle_statu"] != ""]
df_status = df_status.drop_duplicates(subset=["id_statu"]).sort_values("id_statu").reset_index(drop=True)

df_status["id_statu"] = df_status["id_statu"].astype(int)

print(df_status.shape)
df_status.head(10)
```

## Sauvegarde du nettoyage
On sauvegarde le csv dans le dossier 03_gold

```python
df_status.to_csv("../data/03_gold/status.csv", index=False, encoding="utf-8")
```
